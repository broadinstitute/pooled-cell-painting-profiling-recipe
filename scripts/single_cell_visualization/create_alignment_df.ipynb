```python
import os
import pandas as pd
import glob
```

```python
root = '/Users/eweisbar/Desktop/project_folder'
usecols = ['Align_Xshift_DAPI_Painting','Align_Xshift_Phalloidin','Align_Xshift_ZO1','Align_Yshift_DAPI_Painting','Align_Yshift_Phalloidin','Align_Yshift_ZO1']

df_align = pd.DataFrame(columns=usecols)
for i in glob.glob(os.path.join(root, '*/Image.csv')):
    df = pd.read_csv(i)
    df = df[usecols]
    id = i.replace(root,'').replace('Image.csv','').replace('/','')
    df['Metadata_Plate'], df['Metadata_Well'], df['Metadata_Site'] = id.split('-')
    df_align = pd.concat([df_align, df], ignore_index=True)

if not os.path.exists('outputs'):
    os.makedirs('outputs')
df_align.to_csv('outputs/Alignments.csv', index=False)
```
